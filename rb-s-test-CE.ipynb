```python
import rebound
import reboundx

sim = rebound.Simulation()
sim.units = ('AU','yr','Msun')
sim.G = 1 #4*3.141592653589793**2
sim.integrator='whfast'
sim.dt= 10.
# donor star
sim.add(m=1.0, r=0.2)
# companion inside envelope
sim.add(m=0.01, a=0.1)
sim.particles[1].r = 0.1
sim.move_to_com()

rebx = reboundx.Extras(sim)
rl = rebx.load_operator('roche_lobe_mass_transfer')
rebx.add_operator(rl)
rl.params["rlmt_skip_in_CE"] = 0
rl.params['rlmt_donor'] = 0
rl.params['rlmt_accretor'] = 1
sim.particles[0].params['rlmt_Hp'] = 1
sim.particles[0].params['rlmt_mdot0'] = 1e-50
rl.params["rlmt_loss_fraction"] = 0.0
rl.params["jloss_mode"] =0

# Substepping
rl.params["rlmt_substep_max_dm"] = 1e-3
rl.params["rlmt_substep_max_dr"] = 5e-3
rl.params["rlmt_min_substeps"]   = 3
# common-envelope drag parameters
rl.params['ce_rho0'] = 100.
rl.params['ce_alpha_rho'] = -2.0
rl.params['ce_cs'] = 30.0
rl.params['ce_alpha_cs'] = 0.0
rl.params['ce_xmin'] = 0.01
rl.params['ce_Qd'] = 0 #1.0

for itime in range (0,10000,100):
    sim.integrate(itime)
#sim.step()
    print('m0={:.4f}'.format(sim.particles[0].m))
    print('m1={:.4f}'.format(sim.particles[1].m))
    print('a={:.4f}'.format(sim.particles[1].a))
```
